````python
### Warm-ups

# 1. **Middle char(s)**

# * `middle(s)` → If odd length return the middle char, else return
#   the two middle chars.


def middle(s):
    return s[1:-1]
assert middle("abc") == "b"
assert middle("abba") == "bb"
````

````python
# 2. **Title-case safely**

# * `safe_title(s)` → title case but do not lowercase words fully in 
#   ALL CAPS acronyms (e.g., `"learn SQL now"` → `"Learn SQL Now"`;
#   `"use GPU"` → `"Use GPU"`).

# using capwords instead str.title - see https://docs.python.org/3.11/library/stdtypes.html#str.title
from string import capwords

def safe_title(s):
    words = s.split()
    return " ".join([capwords(w) if w.upper() != w else w for w in words])

assert safe_title("learn SQL now") == "Learn SQL Now"
assert safe_title("use GPU") == "Use GPU"
````

````python
# 3. **Reverse words**

# * `reverse_words(s)` → reverse word order but keep internal characters.

import re
from itertools import chain
def reverse_words(s):
    words = s.split()
    spaces = re.findall(r'\s+', s)
    #result = " ".join(reversed(s.split(" ")))  # inverts order of space groups as well
    return "".join(chain(*zip(reversed(words), list(spaces) + [""])))
reverse_words("one two  three")
assert reverse_words("one two  three") == "three two  one"
````

````python
### Core

# 4. **Normalize spaces**

# * `squash_spaces(s)` → collapse multiple spaces/tabs to a single space; strip ends.


def squash_spaces(s):
    return " ".join(s.split())
assert squash_spaces("  a\t b   c ") == "a b c"
````

````python
# 5. **CSV line → list (no quotes)**

# * `split_csv(s)` → split by commas, trim whitespace around fields; 
#   ignore empty trailing field if line ends with comma.


def split_csv(s):
    return [v.strip() for v in s.split(',') if v]
assert split_csv(" a, b ,c,") == ["a","b","c"]
````

````python
# 6. **Mask secrets**

# * `mask_email(s)` → keep first char of user and domain, 
#   mask the rest with `*`, keep TLD.

#    * `"alice@example.com"` → `"a****@e******.com"`


def mask_email(s):
    user, other = s.split('@')
    domain, tld = other.rsplit('.', maxsplit=1)
    user = user[0] + "*" * (len(user) - 1)
    domain = domain[0] + "*" * (len(domain) - 1)
    return user + "@" + domain + "." + tld

mask_email("alice@example.lol.com")
assert mask_email("alice@example.com") == "a****@e******.com"
assert mask_email("alice.bieber@example.lol.com") == "a***********@e**********.com"
````

````python
# 7. **Find all indexes**

# * `find_all(s, sub)` → list of start indices where `sub` occurs (including overlaps).

def find_all(s, sub):
    start = 0
    ind = []
    while start < len(s):
        i = s.find(sub, start)
        if i == -1:
            return ind
        else:
            ind.append(i)
            start = i + 1

assert find_all("aaaa", "aa") == [0,1,2]
assert find_all("aabaaa", "aa") == [0,3,4]
````

````python
# 8. **Anagram check**

# * `is_anagram(a, b)` ignoring spaces, case, punctuation.
# [Anagram](https://en.wikipedia.org/wiki/Anagram)

from string import punctuation  # contains: !"#$%&'()*+,-./:;<=>?@[\]^_`{|}~
def is_anagram(a, b):
    def _clean(s):
        return s.translate(str.maketrans("", "", punctuation)).lower()
    return sorted(_clean(a)) == sorted(_clean(b))

assert is_anagram("Listen", "Silent")
assert is_anagram("Listen!", "'<Silent'>")

````

````python
# 9. **Format table row**

# * `fmt_row(values, widths)` → left-align strings to fixed widths, joined by `" | "`.


def fmt_row(values, widths):
    result = []
    for v, w in zip(values, widths):
        result.append(f"{v:<{w}}")
    return " | ".join(result)

assert fmt_row(["a","bb"], [3,4]) == "a   | bb  "

````

````python
### Challenge

# 10. **Slugify**

# * `slugify(title)` → lowercase, trim, replace runs of non-alnum with 
#   single `-`, remove leading/trailing `-`.

def slugify(title):
    def _clean(s):
        return s.translate(str.maketrans("", "", punctuation)).lower()
    return "-".join(_clean(s) for s in title.split())
    
assert slugify("Hello,  World!!") == "hello-world"


````
