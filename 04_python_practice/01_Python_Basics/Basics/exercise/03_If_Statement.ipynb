````python
### Warm-ups

# 1. **FizzBuzz (classic)**

# * `fizzbuzz(n)` → `"Fizz"`, `"Buzz"`, `"FizzBuzz"`, or `str(n)`.


def fizzbuzz(n):
    fizz = buzz = ""
    if n % 3 == 0:
        fizz = "Fizz"
    if n % 5 == 0:
        buzz = "Buzz"
    fizzbuzz = fizz + buzz
    if not fizzbuzz:
        return str(n)
    return fizzbuzz


assert fizzbuzz(1) == "1"
assert fizzbuzz(3) == "Fizz"
assert fizzbuzz(5) == "Buzz"
assert fizzbuzz(15) == "FizzBuzz"
````

````python
# 2. **Categorize temperature**

# * `temp_label(c)` → `freezing` (<0), `cold` (<10), `mild` (<20), `warm` (<30), `hot` (else).

def temp_label(c):
    if c < 0:
        return  "freezing"
    elif c < 10:
        return "cold"
    elif c < 20:
        return "mild"
    elif c < 30:
        return "warm"
    else:
        return "hot"

assert temp_label(-5) == "freezing"
assert temp_label(10) == "mild"
assert temp_label(40) == "hot"
````

````python
# 3. **Truthy length**

# * `choose_label(s)` → `"empty"` if string is empty/whitespace; `"short"` (≤3), `"long"` (>3).


def choose_label(s):
    if s == "":
        return "empty" 
    if len(s) <= 3:
        return "short"
    return "long"

assert choose_label("") == "empty"
assert choose_label("a") == "short"
assert choose_label("asd") == "short"
assert choose_label("asda") == "long"
````

````python
### Core

#4. **Grade boundaries**

#* `grade(score)` with 0–100 validation; raise `ValueError` otherwise. Return `A/B/C/D/F`.

def grade(score):
    if not (0 <= score <= 100):
        raise ValueError
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    elif score >= 50:
        return "D"
    return "F"

assert grade(0) == "F"
assert grade(49) == "F"
assert grade(50) == "D"
assert grade(70) == "C"
assert grade(80) == "B"
assert grade(90) == "A"
for score in (-5, 102):
    try:
        assert grade(score)
    except ValueError:
        print(f"ValueError raised for {score=}")
````

````python
# 5. **Password strength (simple)**

# * `password_strength(s)` returns `"weak"`, `"medium"`, `"strong"` based on length and presence of digit/special.
# * "weak": less than 6 characters, or only letters or only digits
# * "medium": at least 6 characters, with both letters and digits
# * "strong": at least 8 characters, with letters, digits, and special characters

from string import punctuation, ascii_letters, digits
def password_strength(s):
    #print(s, f"{s.isalnum()=}")
    #print(f"{set(punctuation).intersection(s)=}")
    #print(f"{set(ascii_letters).intersection(s)=}")
    #print(f"{set(digits).intersection(s)=}")
    #print()

    if (
        len(s) >= 8
        and set(punctuation).intersection(s) != set()
        and set(ascii_letters).intersection(s) != set()
        and set(digits).intersection(s) != set()
    ):
        return "strong"

    if (
        len(s) >= 6
        and set(ascii_letters).intersection(s) != set()
        and set(digits).intersection(s) != set()
    ):
        return "medium"
        
    return "weak"

for pw in (
    "abcde",
    "12345",
    "ab!12",
    "abcdea",
    "123451",
    "sdgfkjsgdkfjsgdfk",
    "68172874786827634",
):
    assert password_strength(pw) == "weak"

for pw in (
    "abcde1",
    "12345a",
    "ab!121",
    "sdgfkjsg1fjsgdfk",
    "68172874a86827634",
):
    assert password_strength(pw) == "medium"

    
for pw in (
    "abc!1234",
    "kgs%asg1",
    "68172$874786827634a",
    "asd#fge135",
    "12+3456aasfg",
):
    assert password_strength(pw) == "strong"
````

````python
# 6. **Shipping rules**

# * `shipping_cost(weight_kg, country, premium=False)`:

#    * base by weight: ≤1kg: 5; ≤5kg: 10; else: 20
#    * non-local country adds 15
#    * premium adds +25 at the end

def shipping_cost(weight_kg, country, premium=False, local_country="IN"):
    cost = 0
    
    if weight_kg <= 1:
        cost += 5
    elif weight_kg <= 5:
        cost += 10
    else:
        cost += 20

    if country.upper() != local_country.upper():
        cost += 15
    
    if premium:
        cost += 25

    return cost

for data in (  
    (shipping_cost(0.5, "IN", ), 5),
    (shipping_cost(1.5, "IN", ), 10),
    (shipping_cost(5.1, "IN", ), 20),
    (shipping_cost(10, "GE", ), 35),
    (shipping_cost(10, "GE", local_country="GE", ), 20),
    (shipping_cost(10, "IN", premium=True), 45),
    (shipping_cost(10, "GE", premium=True), 60),
    (shipping_cost(0.5, "GE", premium=False, local_country="GE", ), 5),
):
    assert data[0] == data[1]
````

````python
# 7. **Pick earliest valid date**

# * Given three `(y,m,d)` tuples, `earliest_valid(dates)` where each must pass simple 
# checks (1≤m≤12, 1≤d≤31) else ignore. Return earliest valid or `None`.

from datetime import date
def earliest_valid(dates):
    _dates = []
    for _dt in dates:
        y,m,d = _dt
        try:
            date(y,m,d)
            _dates.append(_dt)
        except ValueError:
            print(f"date {_dt} is invalid and is ignored")
            pass  # tuple could not be converted to date
    
    #print(f"valid dates: {_dates}")
    return sorted(_dates)[0]

assert earliest_valid( [(2025, 5, 1), (2025, 1, 15), (2025, 12, 1), ] ) == (2025, 1, 15)
assert earliest_valid( [(2025, 13, 1), (2025, 12, 32), (2026, 1, 1), ] ) == (2026, 1, 1)
````

````python
# 8. **Safe cast**

# * `to_int(s, default=None)` returns int or default; 
#    only accept pure integer strings (e.g., `" 12 "` ok after strip,
#    `"12.0"` not ok).


def to_int(s, default=None):
    if '.' in s:
        return None
    try:
        return int(s)
    except:
        return None

assert to_int("12") == 12
assert to_int("12.0") == None
assert to_int("12a") == None
````

````python
# 9. **Guard clauses**

# * `discount(price, code)` 
#       → if `price<=0` return 0;
#         if `code=="NONE"` return `price`;
#         if `code=="HALF"` return `price*0.5`;
#         else return `price*0.9`.


def discount(price, code):
    if price <= 0:
        return 0
    elif code in ("NONE", None):
        return price
    elif code == "HALF":
        return price*0.5
    else:
        return price*0.9

assert discount(100, None) == 100.
assert discount(100, "NONE") == 100.
assert discount(100, "HALF") == 50.
assert discount(100, "POSHANKA") == 90.
````

````python
### Challenge

# 10. **Ternary map**

# * `label_numbers(nums)` 
#       → for each n: 
#           `"neg" if n<0`
#           `"zero" if n==0`
#           `"pos" if n>0`.
#  Use a list comprehension with a ternary chain internally (but focus on if-logic correctness).


def label_numbers(nums):
    return [
        "neg" if n<0 else "zero" if n==0 else "pos"
        for n in nums
    ]
assert label_numbers([-1,0,2]) == ["neg","zero","pos"]


````
