````python
# 4) Dictionaries — Exercises

# **Learning goals:** CRUD, iteration patterns, safe access (`get`, `setdefault`), 
#   merging, grouping, transforming, nested access.

### Warm-ups

# 1. **Make profile with defaults**


def make_profile(name, email=None, country="IN"):
    return {
        "name": name,
        "email": email,
        "country": country,
    }
p = make_profile("A","a@example.com")
assert p["name"]=="A" and p["country"]=="IN"
````

````python
# 2. **Safe get path** — nested dict path or default


def get_in(d, path, default=None):
    """Safe get path — nested dict path or default

    Warning: will probably not work safely if input data is structured differently
    """
    data = d
    try:
        for p in path:
            data = data.get(p, default)
    except:
        data = default
    return data

cfg = {"db":{"host":"localhost", "port":5432}}
assert get_in(cfg, ["db","port"]) == 5432
assert get_in(cfg, ["db","user"], "postgres") == "postgres"

assert get_in(cfg, ["db","user","na_nesting_level"], "mydefault") == "mydefault"
assert get_in(cfg, ["db"], "mydefault") == cfg["db"]

````

````python
# 3. **Set with default container**

def multimap_add(mm, key, value):
    """mm: dict key -> list; append value; create list if missing."""
    mm.setdefault(key, []).append(value)
mm = {}
multimap_add(mm,"a",1); multimap_add(mm,"a",2)
assert mm == {"a":[1,2]}
````

````python
### Core

#4. **Invert mapping (values unique)**
#   If duplicate values appear, keep first key.

def invert_unique(d):
    # works but reverses key order
    return dict(zip(reversed(d.values()), reversed(d.keys())))

assert invert_unique({"a":1,"b":2}) == {1:"a",2:"b"}
assert (d2:=invert_unique({"a":1,"c":1,"b":2,})) == {1:"a",2:"b"}

d2
````

````python
def invert_unique_stable(d):
    result = {}
    for k,v in d.items():
        if v in result:
            continue
        result[v] = k
    return result

assert invert_unique_stable({"a":1,"b":2}) == {1:"a",2:"b"}
assert (d2:=invert_unique_stable({"a":1,"c":1,"b":2,})) == {1:"a",2:"b"}

d2
````

````python
# 5. **Invert to sets (values may repeat)**
#    Note: basically multimapp_add - see above

def invert_to_sets(d):
    result = {}
    for k,v in d.items():
        result.setdefault(v, set()).add(k)
    return result    

assert invert_to_sets({"a":1,"b":1,"c":2}) == {1:{"a","b"}, 2:{"c"}}
````

````python
# 6. **Merge with priority** — `override` wins

def merge_priority(base, override):
    result = base.copy()  # Note: only shallow copy !!!
    result.update(override)
    return result    
    
assert merge_priority({"a":1,"b":2},{"b":9,"c":3}) == {"a":1,"b":9,"c":3}
````

````python
# 7. **Group by key function**

def group_by(xs, key_fn):
    result = {}
    for val in xs:
        key = key_fn(val)
        result.setdefault(key, []).append(val)
    return result    

g = group_by(["apple","pear","plum","kiwi"], len)
assert g == {5:["apple"],4:["pear","plum","kiwi"]}
````
